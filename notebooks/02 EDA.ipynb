```python
import numpy as np

import seaborn as sns
import matplotlib.pyplot as plt

from scipy.stats import chi2_contingency
from scipy.stats import ks_2samp

import warnings
warnings.filterwarnings("ignore")

TARGET = "CHURN_VOLUNTARIO"
ID_COL = "CUSTOMER_ID"
DATE_COL = "PERIODO"

df[DATE_COL] = pd.to_datetime(df[DATE_COL])

num_cols = df.select_dtypes(
    include=np.number
).columns.tolist()

cat_cols = df.select_dtypes(
    include=["object","category"]
).columns.tolist()

# excluir target y id

num_cols = [
    c for c in num_cols
    if c != TARGET
]

cat_cols = [
    c for c in cat_cols
    if c != ID_COL
]
```

```python
#1. Cantidad de variables y filas
print("Filas :", len(df))
print("Columnas :", len(df.columns))

print("\nVariables Numéricas")
print(len(num_cols))

print("\nVariables Categóricas")
print(len(cat_cols))
```

```python
#2. Análisis de nulos
nulos = pd.DataFrame({

    "feature":df.columns,

    "nulos":
        df.isnull().sum(),

    "pct_nulos":
        round(
            100*df.isnull().mean(),
            2
        )

})

nulos.sort_values(
    "pct_nulos",
    ascending=False
)
```

```python
#Heatmap de nulos
plt.figure(figsize=(14,6))

sns.heatmap(
    df.isnull(),
    cbar=False
)

plt.show()
```

```python
#Detección MCAR / MAR / MNAR
#Ejemplo para INGRESO_HOGAR.
missing_flag = (
    df["INGRESO_HOGAR"]
    .isnull()
    .astype(int)
)
```

```python
#Cruzar contra variables.
for col in num_cols[:10]:

    corr = np.corrcoef(
        missing_flag,
        df[col].fillna(
            df[col].median()
        )
    )[0,1]

    print(
        col,
        round(corr,3)
    )
```
